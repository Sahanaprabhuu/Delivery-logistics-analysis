```python
import pandas as pd
import sqlite3

# 1. Load the ORIGINAL file fresh (this wipes out any messed-up version of df)
df = pd.read_csv("Delivery_Logistics.csv")

# 2. Turn the fake 1970 dates into real hours (new columns)
df['actual_hours'] = df['delivery_time_hours'].str[-9:].astype(int)
df['expected_hours'] = df['expected_time_hours'].str[-9:].astype(int)

# 3. Flag the repeated IDs and make a new unique ID
df['bad_id_flag'] = df['delivery_id'].duplicated(keep=False).astype(int)
df['clean_id'] = df.index + 1

# 4. Flag the zero-hour deliveries
df['zero_time_flag'] = (df['actual_hours'] == 0).astype(int)

# 5. Calculate delay and the late / failed columns
df['delay_hours'] = df['actual_hours'] - df['expected_hours']
df['is_late'] = (df['delay_hours'] > 0).astype(int)
df['is_failed'] = (df['delivery_status'] == 'failed').astype(int)
df['late_not_failed'] = ((df['is_late'] == 1) & (df['is_failed'] == 0)).astype(int)

# 6. Check everything matches what we found before
print(df.shape)
print(df['bad_id_flag'].sum(), df['zero_time_flag'].sum(), df['is_late'].sum(), df['is_failed'].sum(), df['late_not_failed'].sum())
```

```python
conn = sqlite3.connect("delivery.db")
df.to_sql("deliveries", conn, if_exists="replace", index=False)
pd.read_sql("SELECT COUNT(*) FROM deliveries", conn)
```

```python
pd.read_sql("""
    SELECT delivery_id,COUNT(*) AS TimesUsed
    FROM deliveries
    GROUP BY delivery_id
    HAVING COUNT(*)>1""",conn)
```

```python
pd.read_sql("""
    SELECT delivery_partner, AVG(late_not_failed) AS late_rate
    FROM deliveries
    GROUP BY delivery_partner
    ORDER BY late_rate DESC""",conn)
```

```python
pd.read_sql("""
SELECT region,
       AVG(late_not_failed) AS late_rate
FROM deliveries
GROUP BY region
ORDER BY late_rate DESC
""", conn)
```

```python
pd.read_sql("""
SELECT weather_condition,
       AVG(late_not_failed) AS late_rate
FROM deliveries
GROUP BY weather_condition
ORDER BY late_rate DESC
""", conn)
```

```python
pd.read_sql("""
SELECT weather_condition, clean_id, delivery_partner, delay_hours
FROM (
    SELECT weather_condition, clean_id, delivery_partner, delay_hours,
           ROW_NUMBER() OVER (
               PARTITION BY weather_condition
               ORDER BY delay_hours DESC
           ) AS rn
    FROM deliveries
    WHERE is_failed = 0
)
WHERE rn = 1
""", conn)
```
